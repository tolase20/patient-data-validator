```python
#1. CREATING VARIABLES
Patient_Name = "David Okoro"
Patient_ID = "PT20260801"
Age = 45
Gender = "Male"
Temperature = 37.2
Heart_Rate = 82
Emergency = False

#2. Validating patient age
valid_age = Age >= 1 and Age <= 120

#3. Validating Temperature
valid_temperature = Temperature >= 35 and Temperature <= 42

#4. Validating heart rate
valid_heartrate = Heart_Rate > 40 and Heart_Rate < 200

#5. Validating Patient ID and Name
valid_ID = Patient_ID.strip() != ""
valid_name = Patient_Name.strip() != ""

#6. Validating Gender
if Gender == "Male" or Gender == "Female" or Gender == "Other":
    valid_gender = True
else:
    valid_gender = False

#7. Classifying patient age categories
if Age <= 12:
    age_category = "Child"
elif Age <= 17:
    age_category = "Teenager"
elif Age <= 64:
    age_category = "Adult"
else:
    age_category = "Senior"

#8. Determining overall validation status
valid_record = valid_ID and valid_name and valid_age and valid_gender and valid_heartrate and valid_temperature

print("Patient Record:", "Valid" if valid_record else "Invalid")
```

```python
#9. Creating a function logic ( full validate_patient function)
def validate_patient(name, patient_id, age, gender, temperature, heart_rate):
    results = {}
    results["Name Valid"] = name.strip() != ""
    results["Patient ID Valid"] = patient_id.strip() != ""
    results["Age Valid"] = age >= 1 and age <= 120
    results["Gender Valid"] = gender == "Male" or gender == "Female" or gender == "Other"
    results["Temperature Valid"] = temperature >= 35 and temperature <= 42
    results["Heart Rate Valid"] = heart_rate > 40 and heart_rate < 200
    overall_valid = all(results.values())
    return results, overall_valid


#10. Individual validation functions
def validate_age(value):
    return value >= 1 and value <= 120

def validate_temperature(value):
    return value >= 35 and value <= 42

def validate_heart_rate(value):
    return value > 40 and value < 200


# Call the main function using patient's data
results, overall_valid = validate_patient(Patient_Name, Patient_ID, Age, Gender, Temperature, Heart_Rate)
```

```python
#11. Using a for loop to display results
print("----------------------------------------")
print("VALIDATION RESULTS")
print("----------------------------------------")
for check_name, passed in results.items():
    status = "PASS" if passed else "FAIL"
    print(f"{check_name}: {status}")
```

```python
#14. Generating the final report
def print_report(name, patient_id, age, gender, temperature, heart_rate, emergency):
    results, overall_valid = validate_patient(name, patient_id, age, gender, temperature, heart_rate)

    if age <= 12:
        category = "Child"
    elif age <= 17:
        category = "Teenager"
    elif age <= 64:
        category = "Adult"
    else:
        category = "Senior"

    print("=" * 40)
    print("PATIENT DATA VALIDATION REPORT")
    print("=" * 40)
    print(f"Patient Name: {name}")
    print(f"Patient ID: {patient_id}")
    print(f"Age: {age}")
    print(f"Gender: {gender}")
    print(f"Temperature: {temperature}°C")
    print(f"Heart Rate: {heart_rate} BPM")
    print(f"Emergency Status: {emergency}")
    print("-" * 40)
    print("VALIDATION RESULTS")
    print("-" * 40)
    for check_name, passed in results.items():
        print(f"{check_name}: {passed}")
    print(f"Age Category: {category}")
    print("-" * 40)
    print("FINAL STATUS")
    print("-" * 40)
    print("Patient Record:", "Valid" if overall_valid else "Invalid")
    print("=" * 40)


print_report(Patient_Name, Patient_ID, Age, Gender, Temperature, Heart_Rate, Emergency)
```

```python
#BONUS: Validating multiple patients
patients = [
    {"name": "David Okoro", "id": "PT001", "age": 45, "gender": "Male", "temperature": 37.2, "heart_rate": 82},
    {"name": "Grace Bello", "id": "PT002", "age": 72, "gender": "Female", "temperature": 36.8, "heart_rate": 78},
    {"name": "Samuel Ade", "id": "", "age": 15, "gender": "Male", "temperature": 38.5, "heart_rate": 105},
    {"name": "Anita James", "id": "PT004", "age": 130, "gender": "Female", "temperature": 36.9, "heart_rate": 76},
]

valid_count = 0
invalid_count = 0
invalid_names = []
failure_counts = {}

for p in patients:
    results, overall_valid = validate_patient(p["name"], p["id"], p["age"], p["gender"], p["temperature"], p["heart_rate"])
    print("-" * 40)
    print(f"Patient: {p['name']}")
    for check_name, passed in results.items():
        print(f"  {check_name}: {passed}")
        if not passed:
            failure_counts[check_name] = failure_counts.get(check_name, 0) + 1
    if overall_valid:
        valid_count += 1
    else:
        invalid_count += 1
        invalid_names.append(p["name"])

print("=" * 40)
print("Valid records:", valid_count)
print("Invalid records:", invalid_count)
print("Invalid patients:", invalid_names)
if failure_counts:
    most_common = max(failure_counts, key=failure_counts.get)
    print("Most common failing check:", most_common)
```
